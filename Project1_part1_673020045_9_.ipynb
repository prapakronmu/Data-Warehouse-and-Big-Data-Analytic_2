```python
import json
import re
from collections import Counter
import matplotlib.pyplot as plt
import pandas as pd

file_names = [file1_name, file2_name]

# ฟังก์ชัน Generator อ่านไฟล์ทีละบรรทัด ป้องกัน Error
def iter_jsonl_safe(paths):
    for path in paths:
        if not os.path.exists(path):
            print(f"⚠️ ไม่พบไฟล์: {path}")
            continue
        print(f"📂 กำลังอ่านไฟล์: {path} ...")
        with open(path, 'r', encoding='utf-8', errors='ignore') as f:
            for line in f:
                line_str = line.strip()
                if not line_str:
                    continue
                try:
                    yield json.loads(line_str)
                except Exception:
                    continue
```

```python
total_posts = 0
timestamps = []
languages = Counter()
hashtags = Counter()
sample_posts = []

for post in iter_jsonl_safe(file_names):
    if not isinstance(post, dict):
        continue
    total_posts += 1

    # ดึง Text
    text = post.get('text') or post.get('record', {}).get('text') or ""

    # ดึง Timestamp
    created_at = (post.get('created_at') or post.get('createdAt') or
                  post.get('indexedAt') or post.get('record', {}).get('createdAt'))
    if created_at:
        timestamps.append(created_at)

    # ดึง ภาษา
    langs = post.get('langs') or post.get('record', {}).get('langs') or []
    if isinstance(langs, str):
        langs = [langs]
    elif not isinstance(langs, list):
        langs = []

    if langs:
        for l in langs:
            languages[str(l)] += 1
    else:
        languages['ไม่ระบุ (Unknown)'] += 1

    # ดึง Hashtags และข้อความตัวอย่าง
    if text and isinstance(text, str):
        found_tags = re.findall(r'#\w+', text)
        for tag in found_tags:
            hashtags[tag.lower()] += 1

        if len(sample_posts) < 5 and len(text.strip()) > 30:
            sample_posts.append(text.strip())

print(f"\n✅ อ่านและประมวลผลข้อมูลเสร็จสิ้น! รวมทั้งหมด {total_posts:,} โพสต์")
```

```python
print("=== 1. คนโพสต์เมื่อไร? ===")
print(f"- จำนวนโพสต์รวมทั้งหมด: {total_posts:,} โพสต์")

df_ts = pd.DataFrame()
if timestamps:
    df_ts = pd.DataFrame({'raw_time': timestamps})

    # แก้ไขโดยใส่ utc=True เพื่อแปลงข้อมูลเวลาทุกแบบให้เป็น Datetime มาตรฐานเดียวกัน
    df_ts['datetime'] = pd.to_datetime(df_ts['raw_time'], utc=True, errors='coerce')
    df_ts = df_ts.dropna(subset=['datetime'])

    if not df_ts.empty:
        min_time = df_ts['datetime'].min()
        max_time = df_ts['datetime'].max()
        peak_hour = df_ts['datetime'].dt.hour.mode()
        peak_h_val = peak_hour[0] if not peak_hour.empty else 0

        print(f"- ช่วงเวลาที่ข้อมูลครอบคลุม: ตั้งแต่ {min_time} ถึง {max_time}")
        print(f"- ชั่วโมงที่มีการโพสต์มากที่สุด (UTC): {peak_h_val:02d}:00 - {peak_h_val+1:02d}:00 น.")
```

```python
print("=== 2. ใช้ภาษาอะไร? ===")
print("สัดส่วนภาษาหลักที่พบในชุดข้อมูล:")
for lang, count in languages.most_common(5):
    pct = (count / total_posts * 100) if total_posts > 0 else 0
    print(f"  • {lang}: {count:,} โพสต์ ({pct:.2f}%)")

print("\n=== 3. คนพูดถึงอะไร? ===")
if hashtags:
    print("- Hashtag ยอดนิยม 5 อันดับแรก:")
    for tag, count in hashtags.most_common(5):
        print(f"  • {tag}: {count:,} ครั้ง")
else:
    print("- ไม่พบ Hashtag (#) เด่นชัดในข้อมูลชุดนี้")

print("\n- ตัวอย่างข้อความจริงจากผู้ใช้งาน:")
for i, sample in enumerate(sample_posts[:3], 1):
    clean_s = sample.replace('\n', ' ')
    print(f"  {i}. \"{clean_s[:120]}...\"")
```
